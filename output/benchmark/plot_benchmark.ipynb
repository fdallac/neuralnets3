## **Matrix Multiplication Optimization Benchmark**

```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
mm_methods_list = [
    'Vanilla',
    'LoopUnrolled_8',
    'LoopUnrolled_custom_8',
    'Tiled',
    'OpenMP',
    'SIMD_AVX512',
    'Optimized',
    'OpenBLAS', 
]
```

Load and process benchmark logs

```python
bench = pd.read_csv("matrix_mult_benchmark_logs.csv")
```

```python
# Compute average elapsed times
# TODO drop best and worst times before averaging
avg_times = bench.groupby(['MATRIX_SIZE', 'MM_METHOD'])['ELAPSED_MS'].mean().unstack()
# Reorder columns
avg_times = avg_times[mm_methods_list]
```

### **Benchmark plot**

```python
# Log scales
f = plt.figure(figsize=(10, 6))
f.suptitle('Matrix Multiplication Benchmark', fontweight='bold')
g = f.gca()
avg_times.plot(marker='o', ax=g)
g.set_xlabel('Matrix Size')
g.set_ylabel('Average Elapsed Time (ms)')
g.grid(True, which="both", ls="--")
```

### **Benchmark plot [Log scale]**

```python
# Log scales
f = plt.figure(figsize=(10, 6))
f.suptitle('Matrix Multiplication Benchmark (Log Scale)', fontweight='bold')
g = f.gca()
g.set_yscale('log')
avg_times.plot(marker='o', ax=g)
g.set_xlabel('Matrix Size')
g.set_ylabel('Average Elapsed Time (ms) [Log Scale]')
g.grid(True, which="both", ls="--")
```
